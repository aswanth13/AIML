```python
def dfs(graph, node, goal, visited=None, path=None):

    if visited is None:
        visited = set()

    if path is None:
        path = []

    visited.add(node)
    path.append(node)

    print("Visiting:", node)

    # Goal found
    if node == goal:
        return path

    # Visit each neighbour
    for neighbor in graph.get(node, []):
        if neighbor not in visited:
            result = dfs(graph, neighbor, goal, visited, path)

            if result:
                return result

    # Backtracking
    path.pop()

    return None


# Graph input
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': ['G'],
    'G': []
}

start = 'A'
goal = 'G'

path = dfs(graph, start, goal)

print("Path:", " -> ".join(path))
```
