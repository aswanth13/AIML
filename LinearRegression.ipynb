```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

# Load the dataset
df = pd.read_csv('/content/drive/MyDrive/study_hours_score.csv')

# Select Hours and Score columns and remove missing values
df = df[['Hours', 'Score']].dropna()

# Independent variable (X)
X = df[['Hours']]

# Dependent variable (Y)
Y = df[['Score']]

# Train the model
model = LinearRegression()
model.fit(X, Y)

# Predict values
Y_pred = model.predict(X)

# Plot
plt.figure(figsize=(8,6))
plt.scatter(X, Y, color='blue', label='Data Points')
plt.plot(X, Y_pred, color='red', linewidth=2, label='Regression Line')

plt.title('Simple Linear Regression: Study Hours vs Exam Score')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')

plt.legend()
plt.grid(True)
plt.show()

# Print model parameters
print("Slope (Coefficient):", model.coef_[0][0])
print("Intercept:", model.intercept_[0])

# Calculate R² Score and Mean Squared Error
r2 = r2_score(Y, Y_pred)
mse = mean_squared_error(Y, Y_pred)

print("R² Score:", r2)
print("Mean Squared Error:", mse)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

# Load the dataset
df = pd.read_csv('/content/drive/MyDrive/study_hours_score.csv')

# Select required columns
df = df[['Hours', 'Practice_Tests', 'Score']].dropna()

# Input features (Independent variables)
X = df[['Hours', 'Practice_Tests']]

# Target variable (Dependent variable)
Y = df[['Score']]

# Train multiple linear regression model
model = LinearRegression()
model.fit(X, Y)

# Predict values
Y_pred = model.predict(X)

# Print model parameters
print("Coefficients:", model.coef_[0])
print("Intercept:", model.intercept_[0])

# Evaluation
r2 = r2_score(Y, Y_pred)
mse = mean_squared_error(Y, Y_pred)

print("R² Score:", r2)
print("Mean Squared Error:", mse)


# -------------------------------
# 2D Linear Regression Plot
# -------------------------------

# Keep Practice_Tests constant at its mean value
avg_practice_tests = df['Practice_Tests'].mean()

# Generate range of Hours values
hours_range = np.linspace(df['Hours'].min(), df['Hours'].max(), 100)

# Predict scores using average Practice_Tests
predicted_scores = model.predict(
    pd.DataFrame({
        'Hours': hours_range,
        'Practice_Tests': avg_practice_tests
    })
)

# Plot actual data points
plt.scatter(
    df['Hours'],
    df['Score'],
    color='blue',
    label='Actual Data'
)

# Plot regression line
plt.plot(
    hours_range,
    predicted_scores,
    color='red',
    linewidth=2,
    label='Regression Line'
)

plt.xlabel('Study Hours')
plt.ylabel('Score')
plt.title('Multiple Linear Regression: Score vs Study Hours')
plt.legend()
plt.grid(True)

plt.show()
```
