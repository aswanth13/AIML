```python
import heapq

def uniform_cost_search(graph, start, goal):  #graph → The weighted graph.
                                              #start → Starting node.
                                              #goal → Destination node.
    # Priority queue: (cost, current_node, path)
    queue = [(0, start, [])]    #cost: 0;   current node: start;    path: empty
    visited = set()

    while queue:
        cost, node, path = heapq.heappop(queue) #heapq.heappop() removes the tuple with the smallest costttt.

        if node in visited:
            continue

        visited.add(node)
        path = path + [node]

        if node == goal:
            return cost, path

        for neighbor, edge_cost in graph.get(node, []):
            if neighbor not in visited:
                heapq.heappush(queue, (cost + edge_cost, neighbor, path))

    return float("inf"), []

# Graph input
graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 3), ('C', 2)],
    'C': [('E', 5)],
    'D': [('F', 2) , ('G', 4)],
    'E': [('G', 3)],
    'F': [('G', 1)],
    'G': []
}

start = 'A'
goal = 'G'

cost, path = uniform_cost_search(graph, start, goal)

print("Shortest Cost:", cost)
print("Path:", " -> ".join(path))
```
